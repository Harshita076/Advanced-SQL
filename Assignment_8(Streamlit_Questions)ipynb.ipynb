#Question:- 1 Generate a table generator

```python
# app.py

# Import libraries
import streamlit as st
import pandas as pd

# App title
st.title("Simple Table Generator")

# User input for rows and columns
rows = st.number_input("Enter number of rows", min_value=1, value=3)
cols = st.number_input("Enter number of columns", min_value=1, value=2)

# Create column names automatically
column_names = []

for i in range(cols):
    column_names.append(f"Column {i+1}")

# Create empty table
data = []

for i in range(rows):
    row = []

    for j in range(cols):
        row.append("")

    data.append(row)

# Convert into DataFrame
df = pd.DataFrame(data, columns=column_names)

# Editable table
st.write("Edit Your Table Below")

edited_df = st.data_editor(df)

# Show final table
st.write("Final Table")

st.table(edited_df)
```

###Question:-2 Create a grade calculator or marks between 0 to 100 out of hundred , design the app in such a way that I do not need to input the marks manually using keyboard

```python
# app.py

# Import library
import streamlit as st

# Title
st.title(" Grade Calculator")

st.write("Select your marks using the slider below.")

# Slider (No keyboard typing needed)
marks = st.slider("Choose Marks", 0, 100, 50)

# Show selected marks
st.write("Your Marks:", marks)

# Grade Calculation
if marks >= 90:
    grade = "A+"
elif marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 50:
    grade = "D"
else:
    grade = "Fail"

# Display Grade
st.success(f"Your Grade is: {grade}")
```

###Question:-3 Make a online signing petition for the resignation of Dharmendra Pradhan the education minister of India and create to input boxes one for name in uppercase letter and other fir signature with no specific condition ( but text)

```python
# app.py

# Import library
import streamlit as st

# Title
st.title("Online Petition")

# Petition Heading
st.header("Petition for the Resignation of Dharmendra Pradhan")

st.write("Fill the form below to sign the petition.")

# Name Input (Uppercase)
name = st.text_input("Enter Your Name (UPPERCASE ONLY)")

# Signature Input
signature = st.text_input("Enter Your Signature")

# Button
if st.button("Submit Petition"):

    # Check if name is uppercase
    if name.isupper() and name != "":

        # Check signature is not empty
        if signature != "":

            st.success(" Petition Signed Successfully!")

            st.write("### Submitted Details")
            st.write("Name:", name)
            st.write("Signature:", signature)

        else:
            st.error("Please enter your signature.")

    else:
        st.error("Name must be in UPPERCASE letters only.")
```
