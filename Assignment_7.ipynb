```python
torch.sum(e, dim=0)
```

```python
torch.sum(e)
```

```python
# Mean
torch.sum(e)
#mean along col
torch.mean(e, dim=0)
#mean along row
torch.mean(e, dim=1)
```

```python
# median
torch.median(e)
```

```python
# max and min
torch.max(e)
torch.min(e)
```

```python
# product
torch.prod(e)
```

```python
# standard deviation
torch.std(e)
```

```python
# variance
torch.var(e)
```

```python
# argmax
torch.argmax(e)
```

```python
# argmin
torch.argmin(e)
```

# 4. Matrix operations

```python
a = torch.randint(size=(2,3), low=0, high=10)
c= torch.randint(size=(3,2), low=0, high=10)

print(a)
print(c)
```

```python
# matrix multiplcation
torch.matmul(a, c)
```

```python
vector1 = torch.tensor([1, 2])
vector2 = torch.tensor([3, 4])

# dot product
torch.dot(vector1, vector2)
```

```python
# transpose
torch.transpose(a, 0, 1)
```

```python
b = torch.randint(size=(3,3), low=0, high=10, dtype=torch.float32)
b
```

```python
# determinant
torch.det(b)
```

```python
# inverse
torch.inverse(b)
```

# 5. Comparison operations

```python
i = torch.randint(size=(2,3), low=0, high=10)
j = torch.randint(size=(2,3), low=0, high=10)

print(i)
print(j)
```

```python
# greater than
i > j
# less than
i < j
# equal to
i == j
# not equal to
i != j
# greater that equal to
i>=j
# less than equal to
i<=j
```

# 6. Special functions

```python
h = torch.randint(size=(2,3), low=0, high=10, dtype=torch.float32)
h
```

```python
# log
torch.log(h)
```

```python
# exp
torch.exp(h)
```

```python
# sqrt
torch.sqrt(h)
```

```python
# sigmoid scaling bw 0 and 1
torch.sigmoid(h)
```

```python
# softmax scaling bw 0 and 1 with sum of all elements as 1
torch.softmax(h, dim=0)
```
