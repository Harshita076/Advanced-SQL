```python
# Install Chrome
!wget -q https://dl.google.com/linux/direct/google-chrome-stable_current_amd64.deb
!apt -qq install -y ./google-chrome-stable_current_amd64.deb

# Install required libraries
!pip install -q selenium webdriver-manager

# Import libraries
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

from selenium.webdriver.chrome.options import Options

import time
import pandas as pd

# Chrome options
options = Options()
options.add_argument("--headless=new")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

# Setup driver
driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()),
    options=options
)

# Open Bing
driver.get("https://www.bing.com/")
time.sleep(3)

# Search query
search_box = driver.find_element(By.NAME, "q")
search_box.send_keys("Final")
search_box.send_keys(Keys.RETURN)

time.sleep(5)

# Extract results
results = driver.find_elements(By.CSS_SELECTOR, "li.b_algo")

data = []

for r in results[:5]:
    try:
        title = r.find_element(By.TAG_NAME, "h2").text
        link = r.find_element(By.TAG_NAME, "a").get_attribute("href")
        data.append([title, link])
    except:
        continue

# Print results
print("Top 5 Results:\n")

for i, d in enumerate(data, 1):
    print(f"{i}. {d[0]}")
    print(d[1])
    print()

# Save results
df = pd.DataFrame(data, columns=["Title", "Link"])
df.to_csv("bing_results.csv", index=False)

print("✅ Saved as bing_results.csv")

# Save screenshot
driver.save_screenshot("bing_page.png")
print("📸 Screenshot saved as bing_page.png")

# Close browser
driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.action_chains import ActionChains

import time

# Chrome options
options = Options()
options.add_argument("--headless=new")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

# Start browser
driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()),
    options=options
)

# Open website
driver.get("https://www.bing.com/")
time.sleep(2)

# Search box
search_box = driver.find_element(By.NAME, "q")

# Type search
search_box.send_keys("Selenium Tutorial")
search_box.send_keys(Keys.RETURN)

time.sleep(3)

# Scroll down
driver.execute_script(
    "window.scrollTo(0, document.body.scrollHeight);"
)

print("Scrolled Down")

time.sleep(2)

# Scroll up
driver.execute_script(
    "window.scrollTo(0, 0);"
)

print("Scrolled Up")

time.sleep(2)

# Get page title
print("Title:", driver.title)

# Get current URL
print("URL:", driver.current_url)

# Find multiple results
results = driver.find_elements(By.CLASS_NAME, "b_algo")

print("Total Results:", len(results))

# Print top 5 results
for i, r in enumerate(results[:5], 1):
    print(f"\nResult {i}")
    print(r.text)

# Open new tab
driver.switch_to.new_window('tab')

# Open Google in new tab
driver.get("https://www.google.com")

print("New Tab Opened")

time.sleep(2)

# Switch back to first tab
driver.switch_to.window(driver.window_handles[0])

print("Switched Back")

# Screenshot
driver.save_screenshot("bing_page.png")

print("Screenshot Saved")

# Cookies
cookies = driver.get_cookies()

print("Total Cookies:", len(cookies))

# Refresh page
driver.refresh()

time.sleep(2)

# Find element again
search_box = driver.find_element(By.NAME, "q")

# Mouse action
actions = ActionChains(driver)

actions.move_to_element(search_box).perform()

print("Mouse Hover Done")

# Page source
page = driver.page_source

print("Page Source Length:", len(page))

# Close browser
driver.quit()
```
